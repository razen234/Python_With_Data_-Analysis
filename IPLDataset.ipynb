```python
df = df.drop(columns=['umpire3'])
```

```python
df.shape
```

```python
df['city'] = df['city'].fillna("Unknown")
```

```python
df['winner'].fillna("No Result", inplace=True)
```

```python
df['player_of_match'].fillna("Not Awarded", inplace=True)
```

```python
df['umpire1'].fillna(df['umpire1'].mode()[0], inplace=True)
df['umpire2'].fillna(df['umpire2'].mode()[0], inplace=True)
```

```python
df.to_csv("IPL_Cleaned_Data.csv", index=False)
```
