# Python Lists
### In Python, a list is a built-in data structure that can hold an ordered collection of items. Unlike arrays in some languages, Python lists are very flexible:

### Can contain duplicate items
1. Mutable: items can be modified, replaced, or removed
2. Ordered: maintains the order in which items are added
3. Index-based: items are accessed using their position (starting from 0)
4. Can store mixed data types (integers, strings, booleans, even other lists)
## Creating a List
### Lists can be created in several ways, such as using square brackets, the list() constructor or by repeating elements. Let's look at each method one by one with example:

```python
# Using Square bracket
number = [1,2, 3, 4, 5]
alpha = ['a', 'b', 'c', 'd', 'e']
mixed = [1, 'a', 2, 'b', 3, 'c']
print(number)
print(alpha)
print(mixed)
```

```python
# Using list() constructor
number = list((1,2,3,3,4, 5 ))
alpha = list(('a', 'b', 'c', 'd', 'e'))
mixed = list((1, 'a', 2, 'b', 3, 'c'))
print(number)
print(alpha)
print(mixed)
```

```python
# creating list with repeated elements
repeated = [1, 2, 3] * 3
print(repeated)
```

```python
letters = []

# Repeat 20 times
for i in range(20):
    # Pick a random number between 97 ('a') and 116 (20th letter 't')
    num = random.randint(97,102 )
    
    # Convert number to letter and add it to the list
    letters.append(chr(num))

print(letters)

firstelement = letters[0]
print(firstelement)
```

```python
print("Accending order: ", sorted(letters))
```

```python
print("Decending Order: ",sorted(letters,reverse= True))
```

```python
letters.sort()
print(letters)
```

```python
letters.sort(reverse=True)
print(letters)
```

```python
unique = set(letters)
print(unique)
unique = list(unique)
unique.sort()
print("Accending", unique)
```
