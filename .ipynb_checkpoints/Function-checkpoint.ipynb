```python
def fahrenheit_to_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5.0/9.0
    return celsius

temp_f = float(input("Enter the temperature in Fahrenheit: "))
temp_c = fahrenheit_to_celsius(temp_f)
print(f"{temp_f} degrees Fahrenheit is equal to {temp_c} degrees Celsius.")
```

```python
def area():
    l = int(input("Enter l = ")) # Local variable 
    b = int(input("Enter b = ")) # Local Variable
    a = l*b
    print(a)

area()

i = 1
while i != 0:
    i = int(input("IF You Want to continue press 1 if not 0"))
    if i == 1:                                                                                   
        area()
    elif i == 0:
        break
    else:
        print("Invalid Input Again")
    
```

```python
# Function with argument and return type
def add(a,b):
    return a+b

def sub(a,b):
    return(a-b)

a = int(input("Enter a = "))
b = int(input("Enter b = "))
o = input("Enter o = ")

if o == "+":
    print(add(a,b))

elif o == "-":
    print(sub(a,b))

else:
    print("Invalid please right one")
```

```python
# function with argument and return type 
def area(l,b):
    a = l * b 
    return a 

def volume(l,b,h):
    v = area(l,b)*h
    return v


l = int(input("Enter l = "))
b = int(input("Enter b = "))
ar = area(l,b)
print(f"Area is {ar}")

h = int(input("Enter h = "))
vo = volume(l,b,h)


print(f"Volume is {vo}")
```

```python
infos = ""
def info(name, phone):
    global infos
    a = f"{name}-----{phone}\n"
    infos = infos + a

n = int(input("Enter n = "))
for i in range(n):
    name = input("Enter name = ")
    phone = input("Enter phone")
    info(name,phone)

```

```python
print(infos)
```

# Recursion in Python
### Recursion is a programming technique where a function calls itself either directly or indirectly to solve a problem by breaking it into smaller, simpler subproblems.

```python
def hello():
    print("Hello World")
    hello()
hello()
```

```python
def hello():
    print("Hello World")
    x = input("Enter y for more print")
    if x == 'y':
        hello()

hello()
```
