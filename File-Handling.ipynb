# File Handling in Python
### File handling refers to the proceess of performing operations on a file, such as creating, opening, reading ,writing and closing it through a programming interface. It involves managing the data flow between the program and the file system on the storage device, ensuring that data is handled safely and efficiently.

#### Why do we need File Handling
1. To store data permanently, even after the program ends.
2. To access external files like .txt, .csv, .json, etc.
3. To process large files efficiently without using much memory.
4. To automate tasks like reading configs or saving outputs.

```python
from datetime import datetime

now = datetime.now()
date_string = now.strftime("%y-%m-%d %H:%M")

bill_lines = ""
sub_total = 0

n = int(input("Enter n = "))
for i in range(n):
    name = input("Product Name: ")
    price = int(input("Entert price = "))
    qty = float(input("Enter qty: "))
    total = price * qty

    info = f"{name:^15} {price:^10} {qty:^5} {total:^10.2f}\n"
    bill_lines = bill_lines + info
    sub_total = sub_total + total

vat_rate = 0.13
vat_amount = sub_total * vat_rate
grand_total = sub_total + vat_amount

print("\n" + "="*45)
print("      SAMIRAJ KIRANA PASAL BHAKTAPUR")
print(f"Date: {date_string}")
print("="*45)


print(f"{'Product':<15} {'Price':<10} {'Qty':<5} {'Total':<10}")
print("-" * 45)


print(bill_lines, end="")

print("-" * 45)
# Totals and Tax Breakdown
print(f"{'Sub-Total':<32} Rs. {sub_total:>8.2f}")
print(f"{'VAT (13%)':<32} Rs. {vat_amount:>8.2f}")
print("-" * 45)
print(f"{'GRAND TOTAL':<32} Rs. {grand_total:>8.2f}")
print("="*45)
print("        Thank you for your visit!")


file = open('Bill_Samira_Kirana_Pasal.txt','w')
file.write(f"{'Product':<15} {'Price':<10} {'Qty':<5} {'Total':<10}\n")
file.write(bill_lines)
file.close()

```
