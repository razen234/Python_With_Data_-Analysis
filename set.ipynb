```python
s = set()
s.add("Apple")
s.add("Ball")
s.add("Cat")
s.add("Dog")
print(s)
```

```python
s = {'Ball', 'Apple', 'Dog', 'Cat'}
s.remove("Apple")
print(s)
```

```python
U = {"Ram", "Satya","Bill","Sam","Harry","Sita","Samira","Raj","Steave","William","Samira","Raj"}
ms = {"Ram", "Satya","Bill","Sam","Harry","Sita","Samira","Raj"}
apple = {"Steave","William","Samira","Raj"}
print("Working on both company",ms.intersection(apple))
print("Working on merged company ", ms.union(apple))
```

```python
print("Working only on ms",ms.difference(apple))
print("Working only on apple", apple.difference(ms))
```

```python
print("People who  are unemployed are", U.difference(ms.union(apple)))
```
