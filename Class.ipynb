```python
class Tem:
    def __init__(self, f):
        self.f = f
    
    def tempconv(self):
        c = (self.f-32)*5/9
        return c

f = float(input("Enter the temp in F: "))
obj = Tem(f)
print(f"{f} degree F = {obj.tempconv()} degree C")
```

```python
class Student:
    def __init__(self,name,sub1,sub2,sub3):
        self.name = name
        self.sub1 = sub1
        self.sub2 = sub2
        self.sub3 = sub3
    
    def average(self):
        avg = (self.sub1 + self.sub2 + self.sub3)/3
        print (f"Student Name : {self.name} \t  Average : {avg} ")

obj = Student("Samira",89,95,95)
obj.average()
```

```python
class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks
    
    def average(self):
        sum = 0 
        for i in range(len(self.marks)):
            sum = sum + self.marks[i]
        
        print(sum)
        avg = sum /3
        print (f"Student Name : {self.name} \t  Average : {avg} ")

obj = Student("tony stark",[89,95,95])
obj.average()

obj.name = "ironman"
obj.average()

```

# Important
### 1. Abstraction: 
#### Hiding the implemenation details of a class and only showing the essential features to the user 
### 2. Encapsulation
#### wrapping data and functions into a single unit(object)


```python
class Area:
    def __init__(self,l,b):
        self.l = l
        self.b = b
    def area(self):
        a = self.l*self.b
        print(a)

class Volume:
    def __init__(self,l,b,h):
        self.l = l
        self.b = b
        self.h = h
    def volume(self):
        v = self.l * self.h * self.b
        print(v)


obj1 = Area(10,5)
obj2 = Volume(10,5,2)

obj1.area()
obj2.volume()
```
