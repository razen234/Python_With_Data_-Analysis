```python
#Login System
username = "admin"
password = "admin"
count = 1
def login():
    global count
    try:
        user = input("Enter user name: ")
        pas = input("Enter password: ")
        assert (username == user and password == pas)
        print("Welcome to Page")

    except AssertionError:
        print("The username and password does not match")
        count = count+1
        login()
        

login()
print(f"Attemped:  {count} Times")
```

```python
username = "admin"
password = "admin"
history = [] # A list to track attempt status

def login():
    try:
        user = input("Enter user name: ")
        pas = input("Enter password: ")
        
        if username == user and password == pas:
            print("Welcome to Page")
            history.append("success")
        else:
            history.append("failed")
            raise ValueError
            
    except ValueError:
        print("The username and password does not match")
        login()

login()

# Using the .count() method on your history list
total_attempts = len(history)
failed_count = history.count("failed") 

print(f"Total attempts: {total_attempts}")
print(f"You failed {failed_count} times before logging in.")
```

```python
username = "Samira"
password = "1234"
count = 5

def login():
    global count

    if count == 0:
        print("Not Allowed")
        return

    user = input("Enter user name: ")
    pas = input("Enter password: ")

    if user == username and pas == password:
        print("Welcome to Page")
        return
    else:
        count -= 1
        print("Incorrect! Attempts left:", count)
        login()

login()
```

```python
#ATM Withdrawal Handling 
balance = 5000

while True:
    try:
        amount = float(input("Enter withdrawal amount: "))

        if amount > balance:
            raise ValueError ("Insufficient Balance")

        balance = balance - amount
        print("Transaction Successful Rs.",amount)
        print("Balance:", balance)

    except ValueError:
        print("Invalid input enter number only")
    

    choice = input("Press 1 to continue, 0 to exit: ")

    if choice == '0':
        print("Thank you")
        break
    elif choice != '1':
        print("Invalid choice, exiting")
        break

    if balance == 0:
        print("Balance is 0")
        break

```
