# Python Dictonary 
### A python dictonary is a data structure that stores information in key-valye pairs. While keys must be unique and immutable like stings or numbers, the values can be of any data type, whether mutable or immutable. This makes dictionaries ideal for accessing data by a specific name rather than a numeric postion like in list.

```python
data = {"Name": "Samira", "Age": 22}
print(data)
print(type(data))
```

# Creating a Dictionary
### A dictonary is created by writing key-value pairs inside {}, where each key is connected to a value using colon (:). A dictonary can also be created using the dict() function.

```python
d1 = {1:'Samira', 2: 'For', 3:'Raj'}
print(d1)

# using dict() constructor
d2 = dict(a = "Hello", b = "Samira" , c = "Raj" )
print(d2)
```

# Accessing Dictionary Items
## A value in a dictionary is accessed by using its key. This can be done either with square brackets [ ] or with the get() method. Both return the value linked to the given key.

```python
# Read  
a = {'Raj': '6380102281', 'Samira': '9862478815'}
print(a)
print(a.values())
print(a.keys())
print(a.items())
```

```python
for i in a:
    print(i)
```

# Iterating Through a Dictonary 
### A dictonary can be traversed using a for loop to access its key, values or both key-value pairs by using the build-in methods key(), values(), and items()


```python
d = {'Raj': '6380102281', 'Samira': '9862478815'}
# Access using key 
print(d["Raj"])

# Access using get()
print(d.get("Raj"))

print(d.get("S"))


```

```python
d = {'Raj': '6380102281', 'Samira': '9862478815'}

# Iteration over keys
for key in d:
    print(key)
# Iterantion over values 
for values in d.values():
    print(values)
# Iteration over key - values pairs
for key,values in d.items():
    print(f"{key}: {values}")
```

```python
#Update
d = {'Raj': '6380102281', 'Samira': '9862478815'}
d["Raj"] = '9862478815'
d
```

## Removing Dictionary Items
### Dictionary items can be removed using built-in deletion methods that work on keys:
1. del: removes an item using its key
2. pop(): removes the item with the given key and returns its value
3. clear(): removes all items from the dictionary
4. popitem(): removes and returns the last inserted key–value pair

```python
d = {'Raj': '6380102281', 'Samira': '9862478815','Shreyna':'9850000600'}
del d["Raj"]
print(d)
```

```python
# using pop
val = d.pop("Samira")
print(val)
print(d)
```

```python
#using popitem
d = {'Raj': '6380102281', 'Samira': '9862478815','Shreyna':'9850000600'}
key, val = d.popitem()
print(f"key:{key}, Value: {val}")
```

```python
d = {'Raj': '6380102281', 'Samira': '9862478815','Shreyna':'9850000600'}
d.clear()
print(d)
```

# Nested Dictonaries
### A nested dicitonary is a dictionary that contain another dictonary as one of its value.

```python
d = {'Ram':[9862478815,9805674711],
    'Shyam':[9825266728,9805745997],
    'Hari':[6380102281,85678953214]}
d
```

```python
d = dict()
n = int(input("Entert Number of input: "))
for i in range(n):
    Name = input("Enter user name: ")
    sim1 = int(input("Enter sim1 number: "))
    sim2 = int(input("Enter sim2 number: "))
    d[Name] = [sim1, sim2]
print(d)
```
